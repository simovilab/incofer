# Revisión del _feed_ de INCOFER

```python
import pandas as pd
import geopandas as gpd
from shapely.geometry import LineString
```

```python
agency = pd.read_csv('files/agency.txt')
calendar_dates = pd.read_csv('files/calendar_dates.txt')
calendar = pd.read_csv('files/calendar.txt')
feed_info = pd.read_csv('files/feed_info.txt')
routes = pd.read_csv('files/routes.txt')
shapes = pd.read_csv('files/shapes.txt')
stop_times = pd.read_csv('files/stop_times.txt')
stops = pd.read_csv('files/stops.txt')
trips = pd.read_csv('files/trips.txt')
```

```python
geostops = gpd.GeoDataFrame(
    stops,
    geometry=gpd.points_from_xy(stops.stop_lon, stops.stop_lat),
    crs="EPSG:4326",
)
```

```python
geoshapes = gpd.GeoDataFrame(
    shapes.sort_values("shape_pt_sequence")
    .groupby("shape_id")
    .apply(lambda x: LineString(zip(x["shape_pt_lon"], x["shape_pt_lat"])))
    .rename("geometry")
)
```

```python
# Visualize all stops in an interactive map (avoids matplotlib backend issues)
geostops.explore()
```
